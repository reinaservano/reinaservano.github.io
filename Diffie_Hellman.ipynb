# Diffie-Hellman Key Exchange Lab

In this example, you'll learn how the Diffle-Hellman Key Exchange works.

Let's imagine Alice and Bob trying to exchange secret messages without Eve interfering. 

First, Alice and Bob agree on using a public number and another public number to mod. In this example, they agree on using 3 and 17.

**3** mod **17**

Then, Alice chooses a random private number.

3 **^15** mod 17 = ***b***

Bob also selects his random private number.

3 **^13** mod 17 = ***12***

Now, Eve can interfere and get b & 12 but not the random private number.

Alice then takes Bob's public result.

**12** ^ 15 mod 17 = ***10***

Bob does the vice-versa.

**b** ^13 mod 17 = ***10***

Without 15 or 13, Eve cannot find the solution. Therefore, it is practically impossible to crack easily.
